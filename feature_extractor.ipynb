```python
import torch
import os
import pandas as pd
from utils import extract_features
```

```python
print(f"Is CUDA available? {torch.cuda.is_available()}")
print(f"GPU Name: {torch.cuda.get_device_name(0) if torch.cuda.is_available() else 'None'}")
device = torch.device("cuda" if torch.cuda.is_available() else "cpu")
```

```python
features = []
labels = []
```

```python
for label in ['NORMAL', 'PNEUMONIA']:
    folder = os.path.join('chest_xray/train/', label)
    for file in os.listdir(folder):
        if not file.endswith(".jpeg"):
            continue
        img_path = folder + '/' + file
        feat = extract_features(img_path)
        features.append(feat)
        labels.append(label)
```

```python
df_f_train = pd.DataFrame(features)
df_l_train = pd.DataFrame(labels)

df_train = pd.merge(df_f_train, df_l_train, left_index=True, right_index=True)
```

```python
df_train.head()
```

```python
df_train['0_y'].value_counts()
```

```python
df_train['0_y'].replace({'NORMAL': 0, 'PNEUMONIA': 1}, inplace=True)
df_train['0_y'].astype("int64")
```
