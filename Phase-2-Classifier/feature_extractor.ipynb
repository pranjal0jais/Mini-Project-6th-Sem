```python
import os
import pandas as pd
from utils.image_utils import extract_features
from img2vec_pytorch import Img2Vec
```

```python
img2vec = Img2Vec(model="densenet121", cuda=True)
```

```python
def convert_to_vector(folder_path, file_name):
    features = []
    labels = []
    for file in os.listdir(folder_path):
        print("Processing " + file)
        if not file.endswith(".jpeg"):
            continue
        img_path = folder_path + '/' + file
        feat = extract_features(img2vec, img_path)
        if file.__contains__("bacteria"):
            label = 0
        else:
            label = 1
        features.append(feat)
        labels.append(label)

    df_f = pd.DataFrame(features)
    df_l = pd.DataFrame(labels, columns=["label"])
    df = pd.concat([df_f, df_l], axis=1)
    output_path =os.getcwd() +  "\data\\" + file_name.strip() + ".csv"

    df.to_csv(output_path, index=False)

    print_summary(df, output_path)
```

```python
def print_summary(df, output_path):
    print("\n===== DATA SUMMARY =====")
    print("Total samples       :", len(df))
    print("Bacteria samples(0) :", (df["label"] == 0).sum())
    print("Virus samples(1)    :", (df["label"] == 1).sum())
    print("Feature vector size :", df.shape[1]-1)
    print("DataFrame shape     :", df.shape)
    print("Saved to            :", output_path)
```

```python
convert_to_vector("../chest_xray/train/PNEUMONIA", "train")
```

```python
convert_to_vector("../chest_xray/test/PNEUMONIA", "test")
```
